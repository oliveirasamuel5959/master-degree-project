```python
data_loader = DataLoader(path=RLVS_INPUT_DIR)
rlvs_videos_files = data_loader.get_video_files()

print(f"Found {len(rlvs_videos_files)} video files in {RLVS_INPUT_DIR}")
print(f"Video files: {[video.name for video in rlvs_videos_files]}")
```

```python
def display_videos_grid(video_files, columns=4):
    html = f"""
    <div style="
        display:grid;
        grid-template-columns:repeat({columns}, 1fr);
        gap:0px;
        align-items:start;
    ">
    """

    for video_path in video_files:
        html += f"""
        <video
            autoplay
            muted
            loop
            playsinline
            controls
            style="
                width:100%;
                display:block;
                margin:0;
                padding:0;
            "
        >
            <source src="{video_path}" type="video/mp4">
        </video>
        """

    html += "</div>"

    display(HTML(html))
```

#### Display Real Life Violence Situations Dataset

```python
display_videos_grid(rlvs_videos_files, columns=8)
```
