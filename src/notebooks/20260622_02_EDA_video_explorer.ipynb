```python
ROOT_DIR = Path.cwd().parent.parent
sys.path.insert(0, str(ROOT_DIR / "src"))

print(ROOT_DIR)
```

```python
OUTPUT_DIR = ROOT_DIR / "outputs" / "notebooks" / "20260622-02-eda-video"
OUTPUT_DIR.mkdir(exist_ok=True)

UCF_CRIME_INPUT_DIR = ROOT_DIR / "data" / "UCF_Crime" / "raw"
```

```python
data_sample = []
videos_per_class = 4

classes = [d for d in Path(UCF_CRIME_INPUT_DIR).iterdir() if d.is_dir()]

for class_dir in classes:
    videos = list(class_dir.glob("*.mp4"))[:videos_per_class]
    data_sample.extend(videos)

print(f"Found {len(data_sample)} video files in {UCF_CRIME_INPUT_DIR}")
print(f"Video files: {[video.name for video in data_sample]}")
    
```

```python
def display_videos_grid(video_files, columns=4):
    html = f"""
    <div style="
        display:grid;
        grid-template-columns:repeat({columns}, 1fr);
        gap:0px;
        align-items:start;
    ">
    """

    for video_path in video_files:
        html += f"""
        <video
            autoplay
            muted
            loop
            playsinline
            controls
            style="
                width:100%;
                display:block;
                margin:0;
                padding:0;
            "
        >
            <source src="{video_path}" type="video/mp4">
        </video>
        """

    html += "</div>"

    display(HTML(html))
```

#### Display Real Life Violence Situations Dataset

```python
display_videos_grid(data_sample, columns=8)
```
